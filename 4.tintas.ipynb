## Problema das tintas

```python
%pip install pulp -q

import pulp

prob = pulp.LpProblem("Tintas", pulp.LpMinimize)

# Variaveis para tinta SR (1000 litros)
A_sr = pulp.LpVariable("SolA_SR", lowBound=0, cat='Continuous')
B_sr = pulp.LpVariable("SolB_SR", lowBound=0, cat='Continuous')
SEC_sr = pulp.LpVariable("SEC_SR", lowBound=0, cat='Continuous')
COR_sr = pulp.LpVariable("COR_SR", lowBound=0, cat='Continuous')

# Variaveis para tinta SN (250 litros)
A_sn = pulp.LpVariable("SolA_SN", lowBound=0, cat='Continuous')
B_sn = pulp.LpVariable("SolB_SN", lowBound=0, cat='Continuous')
SEC_sn = pulp.LpVariable("SEC_SN", lowBound=0, cat='Continuous')
COR_sn = pulp.LpVariable("COR_SN", lowBound=0, cat='Continuous')

## RESTRICOES
# Restricoes de qtd total
prob += A_sr + B_sr + SEC_sr + COR_sr == 1000 # SR
prob += A_sn + B_sn + SEC_sn + COR_sn == 250 # SN

# Restricoes de composicao - SR
prob += 0.3*A_sr + 0.6*B_sr + 1.0*SEC_sr >= 250 # 25% SEC de 1000
prob += 0.7*A_sr + 0.4*B_sr + 1.0*COR_sr >= 500 # 50% COR de 1000

# Restricoes de composicao - SN
prob += 0.3*A_sn + 0.6*B_sn + 1.0*SEC_sn >= 50 # 20% SEC de 250
prob += 0.7*A_sn + 0.4*B_sn + 1.0*COR_sn >= 125 # 50% COR de 250

# Funcao obj: minimizar custo
prob += 1.5*(A_sr + A_sn) + 1.0*(B_sr + B_sn) + 4.0*(SEC_sr + SEC_sn) + 6.0*(COR_sr + COR_sn)

prob.solve(pulp.PULP_CBC_CMD(msg=False))

# Resultados
print("\nPROBLEMA DAS TINTAS - CUSTO MÍNIMO\n")
print(f"Custo Total Mínimo: R$ {pulp.value(prob.objective):.2f}\n")

print("Totais de insumos a serem comprados:")
print(f"- SolA total: {A_sr.varValue + A_sn.varValue:.2f} litros")
print(f"- SolB total: {B_sr.varValue + B_sn.varValue:.2f} litros")
print(f"- SEC total:  {SEC_sr.varValue + SEC_sn.varValue:.2f} litros")
print(f"- COR total:  {COR_sr.varValue + COR_sn.varValue:.2f} litros")
```

## Problema da Clique Máxima

```python
%pip install pulp -q

import pulp

prob = pulp.LpProblem("CliqueMaxima", pulp.LpMaximize)

# ENUNCIADO
# Dado um grafo G = (V, E), determinar a clique maxima de G,
# ou seja, o maior subgrafo completo em G.

# ENTRADA: grafo simples G(V, E)
V = [1, 2, 3, 4, 5, 6]
E = [
  (1, 2), (1, 3), (1, 4),
  (2, 3), (2, 4),
  (3, 4), (3, 5),
  (4, 6), (5, 6),
]

# Conjunto de arestas para consulta rapida
E_set = set()
for u, v in E:
  E_set.add((u, v))
  E_set.add((v, u))

# VARIAVEIS DE DECISAO
# x_v = 1, se v pertencer a clique maxima; x_v = 0, se nao
# para todo v pertencente a V
x = {v: pulp.LpVariable(f"x_{v}", cat='Binary') for v in V}

## RESTRICOES
# (dominio) x_v pertence a {0, 1} — garantido por cat='Binary'

# (clique) x_v + x_u <= 1, para todo (u, v) nao pertencente a E
for i in range(len(V)):
  for j in range(i + 1, len(V)):
    u, v = V[i], V[j]
    if (u, v) not in E_set:
      prob += x[u] + x[v] <= 1

# FUNCAO OBJETIVO: max somatorio de x_v, v pertencente a V
prob += pulp.lpSum(x[v] for v in V)

prob.solve(pulp.PULP_CBC_CMD(msg=False))

# Resultados
escolha = {v: x[v].varValue for v in V}
clique = [v for v in V if escolha[v] == 1]
tamanho = int(pulp.value(prob.objective))

print("\nPROBLEMA DA CLIQUE MAXIMA\n")
print(f"Tamanho da clique maxima: {tamanho}\n")
print(f"Vertices da clique: {clique}\n")

print("Participacao de cada vertice:")
print("Vertice\tNa clique?")
for v in V:
  marca = 'sim' if escolha[v] == 1 else 'nao'
  print(f"{v}\t{marca}")

print("\nVerificacao da clique:")
pares_clique = [(u, v) for i, u in enumerate(clique) for v in clique[i + 1:]]
pares_faltantes = [(u, v) for u, v in pares_clique if (u, v) not in E_set]
if pares_faltantes:
  print("Aviso: pares sem aresta encontrados:", pares_faltantes)
else:
  print("Todos os pares de vertices selecionados estao conectados por arestas em E.")
```

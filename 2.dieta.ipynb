## Problema da dieta

```python
%pip install pulp -q

import pulp

prob = pulp.LpProblem("Dieta", pulp.LpMinimize)

# Variaveis para cada ingrediente (não negativas)
x1 = pulp.LpVariable("Ingrediente1", lowBound=0, cat='Continuous')
x2 = pulp.LpVariable("Ingrediente2", lowBound=0, cat='Continuous')
x3 = pulp.LpVariable("Ingrediente3", lowBound=0, cat='Continuous')
x4 = pulp.LpVariable("Ingrediente4", lowBound=0, cat='Continuous')
x5 = pulp.LpVariable("Ingrediente5", lowBound=0, cat='Continuous')
x6 = pulp.LpVariable("Ingrediente6", lowBound=0, cat='Continuous')

# Restricoes de vitaminas
prob += 1*x1 + 0*x2 + 2*x3 + 2*x4 + 1*x5 + 2*x6 >= 9 # Vitamina A
prob += 0*x1 + 1*x2 + 3*x3 + 1*x4 + 3*x5 + 2*x6 >= 19 # Vitamina C

# Funcao obj: min custo
prob += 35*x1 + 30*x2 + 60*x3 + 50*x4 + 27*x5 + 22*x6

prob.solve(pulp.PULP_CBC_CMD(msg=False))

# Pegando os resultados
qtd = [x1.varValue, x2.varValue, x3.varValue, x4.varValue, x5.varValue, x6.varValue]
custos = [35, 30, 60, 50, 27, 22]
custo_total = sum(qtd[i] * custos[i] for i in range(6))

# Calcular vitaminas ingeridas
vitA = 1*qtd[0] + 0*qtd[1] + 2*qtd[2] + 2*qtd[3] + 1*qtd[4] + 2*qtd[5]
vitC = 0*qtd[0] + 1*qtd[1] + 3*qtd[2] + 1*qtd[3] + 3*qtd[4] + 2*qtd[5]

# Resultados
print("\nPROBLEMA DA DIETA - CUSTO MÍNIMO\n")
print(f"Custo Total Mínimo: R$ {custo_total:.2f}\n")
print("Composição Ótima:")
print("Ingrediente\tQuantidade\tCusto Unitário\tCusto Parcial")
for i in range(6):
  print(f"{i+1}\t\t{qtd[i]:.2f}\t\t{custos[i]}\t\t{custos[i]*qtd[i]:.2f}")

print("\nAtendimento das Vitaminas:")
print(f"Vitamina A: {vitA:.2f} (mínimo 9)")
print(f"Vitamina C: {vitC:.2f} (mínimo 19)")
```

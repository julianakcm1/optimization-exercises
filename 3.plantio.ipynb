## Problema do plantio

```python
%pip install pulp -q

import pulp

prob = pulp.LpProblem("Plantio", pulp.LpMaximize)

# Variaveis
M = pulp.LpVariable("Milho", lowBound=0, cat='Continuous')
A = pulp.LpVariable("Arroz", lowBound=0, cat='Continuous')
F = pulp.LpVariable("Feijao", lowBound=0, cat='Continuous')

## RESTRICOES
# Restricoes de area maxima por cultura
prob += M <= 660
prob += A <= 880
prob += F <= 400

# Restricao de area total
prob += M + A + F <= 1400

# Restricoes de agua por fazenda (apos multiplicar)
prob += 5.5*M + 4*A + 3.5*F <= 6300 # fazenda 1
prob += 5.5*M + 4*A + 3.5*F <= 4738.46 # fazenda 2
prob += 5.5*M + 4*A + 3.5*F <= 3800 # fazenda 3

# Funcao obj: max lucro
prob += 5000*M + 4000*A + 1800*F

prob.solve(pulp.PULP_CBC_CMD(msg=False))

# resultados
milho = M.varValue
arroz = A.varValue
feijao = F.varValue
lucro_total = 5000*milho + 4000*arroz + 1800*feijao

print("\nPROBLEMA DO PLANTIO - LUCRO MÁXIMO\n")
print(f"Lucro Total Máximo: R$ {lucro_total:.2f}\n")
print("Áreas Totais por Cultura:")
print(f"- Milho:  {milho:.2f} acres")
print(f"- Arroz:  {arroz:.2f} acres")
print(f"- Feijão: {feijao:.2f} acres")
```

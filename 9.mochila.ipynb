## Problema da mochila

```python
%pip install pulp -q

import pulp

prob = pulp.LpProblem("Mochila", pulp.LpMaximize)

# Um aventureiro dispoe de uma mochila com capacidade maxima de 15 kg.
# Ha 7 itens candidatos, cada um com peso (kg) e valor (utilidade).
# Objetivo: escolher os itens que maximizam o valor total transportado
# sem ultrapassar a capacidade da mochila. Cada item pode ser levado no maximo uma vez.

# Variaveis binarias: 1 se o item for levado, 0 caso contrario
x1 = pulp.LpVariable("Item1", cat='Binary')
x2 = pulp.LpVariable("Item2", cat='Binary')
x3 = pulp.LpVariable("Item3", cat='Binary')
x4 = pulp.LpVariable("Item4", cat='Binary')
x5 = pulp.LpVariable("Item5", cat='Binary')
x6 = pulp.LpVariable("Item6", cat='Binary')
x7 = pulp.LpVariable("Item7", cat='Binary')

# Pesos (kg) e valores de cada item
pesos   = [2, 3, 4, 5, 6, 7, 9]
valores = [10, 13, 18, 20, 24, 28, 35]
capacidade = 15

## RESTRICAO
# Restricao de capacidade da mochila
prob += 2*x1 + 3*x2 + 4*x3 + 5*x4 + 6*x5 + 7*x6 + 9*x7 <= capacidade

# Funcao obj: maximizar o valor total dos itens levados
prob += 10*x1 + 13*x2 + 18*x3 + 20*x4 + 24*x5 + 28*x6 + 35*x7

prob.solve(pulp.PULP_CBC_CMD(msg=False))

# Pegando os resultados
escolha = [x1.varValue, x2.varValue, x3.varValue, x4.varValue, x5.varValue, x6.varValue, x7.varValue]
valor_total = sum(escolha[i] * valores[i] for i in range(7))
peso_total = sum(escolha[i] * pesos[i] for i in range(7))

print("\nPROBLEMA DA MOCHILA - VALOR MÁXIMO\n")
print(f"Valor Total Máximo: {valor_total:.0f}")
print(f"Peso Total Utilizado: {peso_total:.0f} / {capacidade} kg\n")

print("Itens Selecionados:")
print("Item\tLevar?\tPeso (kg)\tValor")
for i in range(7):
  marca = 'sim' if escolha[i] == 1 else 'não'
  print(f"{i+1}\t{marca}\t{pesos[i]}\t\t{valores[i]}")

print(f"\nUso da Capacidade: {peso_total/capacidade*100:.2f}%")
```

## Problema da mochila

```python
%pip install pulp -q

import pulp

prob = pulp.LpProblem("Mochila", pulp.LpMaximize)

# ENUNCIADO
# Dado um conjunto de n itens e um inteiro W que representa a capacidade da mochila.
# Cada item i possui um valor v_i e um peso w_i.
# Determinar o subconjunto de itens que maximiza o somatorio dos valores,
# respeitando a capacidade de peso da mochila.

# ENTRADA: n itens com pesos w_i e valores v_i, e capacidade W
n = 7
W = 15
pesos = [2, 3, 4, 5, 6, 7, 9]
valores = [10, 13, 18, 20, 24, 28, 35]
I = list(range(1, n + 1))

# VARIAVEIS DE DECISAO
# x_i = 1, se o item i estiver na mochila; x_i = 0, caso contrario
# para todo i pertencente a {1, 2, ..., n}
x = {i: pulp.LpVariable(f"x_{i}", cat='Binary') for i in I}

## RESTRICOES
# (dominio) x_i pertence a {0, 1} — garantido por cat='Binary'

# (capacidade) somatorio de w_i * x_i <= W, para i pertencente a {1, 2, ..., n}
prob += pulp.lpSum(pesos[i - 1] * x[i] for i in I) <= W

# FUNCAO OBJETIVO: max somatorio de v_i * x_i, para i pertencente a {1, 2, ..., n}
prob += pulp.lpSum(valores[i - 1] * x[i] for i in I)

prob.solve(pulp.PULP_CBC_CMD(msg=False))

# Resultados
escolha = {i: x[i].varValue for i in I}
valor_total = pulp.value(prob.objective)
peso_total = sum(escolha[i] * pesos[i - 1] for i in I)
itens_selecionados = [i for i in I if escolha[i] == 1]

print("\nPROBLEMA DA MOCHILA - VALOR MÁXIMO\n")
print(f"Valor Total Máximo: {valor_total:.0f}")
print(f"Peso Total Utilizado: {peso_total:.0f} / {W} kg\n")

print("Itens Selecionados:")
print("Item\tLevar?\tPeso (kg)\tValor")
for i in I:
  marca = 'sim' if escolha[i] == 1 else 'não'
  print(f"{i}\t{marca}\t{pesos[i - 1]}\t\t{valores[i - 1]}")

print(f"\nUso da Capacidade: {peso_total / W * 100:.2f}%")
```
